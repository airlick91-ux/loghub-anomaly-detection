```python
import joblib
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score
```

```python
matrix_df = pd.read_csv("Event_occurrence_matrix.csv")
label_df = pd.read_csv("anomaly_label.csv")
```

```python
matrix_df
```

```python
print("Reussite de chargement des datasets")
print(f"Dimensions de la matrice: {matrix_df.shape}")
print(f"Dimensions du label: {label_df.shape}")
```

```python
df = pd.merge(matrix_df, label_df, on="BlockId")
```

```python
print(df.columns)
```

```python
colonnes_evenements = [f'E{i}' for i in range(1, 30)]
X = df[colonnes_evenements]
Y = df["Label_y"]
```

```python
Y
```

```python
if Y.dtype == object:
    Y = Y.apply(lambda val: 0 if str(val).lower() in ['success', 'normal', '0', 'false'] else 1)
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
modele = DecisionTreeClassifier(random_state=42, max_depth=10)
modele.fit(x_train, y_train)
```

```python
Y_pred = modele.predict(x_test)
```

```python
print(f"Précision globale (Accuracy) : {accuracy_score(y_test, Y_pred):.4f}\n")
print("Rapport de classification :")
print(classification_report(y_test, Y_pred))
```

```python
joblib.dump(modele, "arbre_de_decision_HDFS_v1.pkl")
```
